```python

API_TOKEN = "your_actual_token_here" # https://radogost.dariah.pl/dataverseuser.xhtml?selectTab=apiTokenTab
PID    = "doi:10.82343/IJSS86" # zmień
SUFFIX = PID.split("/")[-1]          # nazwa sufix to końcówka doi, np. IJSS86

DRAFT_JSON = f"{SUFFIX}_draft.json"
README_OUT = f"README_{SUFFIX}.txt"

```



# Krok 1 : zapisz metadane w kroku 1

```python
import subprocess, sys

result = subprocess.run(
    [
        sys.executable,
        "Radogost_FileList/utils/dataverse_metadata_radogost.py",
        "--mode",      "draft",
        "--api-token", API_TOKEN,
        "--pid",       PID,
        "--output",    DRAFT_JSON,
    ],
    capture_output=True, text=True, encoding="utf-8"
)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
```

# Krok 2 : zapisz plik readme

```python
result = subprocess.run(
    [
        sys.executable,
        "Radogost_FileList/utils/generate_readme.py",
        "--input",  DRAFT_JSON,
        "--output", README_OUT,
    ],
    capture_output=True, text=True, encoding="utf-8"
)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
```

```python
from google.colab import files
import os

for fpath in [DRAFT_JSON, README_OUT]:
    if os.path.exists(fpath):
        print(f"Downloading {fpath}  ({os.path.getsize(fpath):,} bytes)")
        files.download(fpath)
    else:
        print(f"Not found: {fpath} — run previous cells first")
```
